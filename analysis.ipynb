```python
import pandas as pd
from typing import Dict
```

```python
def analyze_metrics(csv_path: str) -> Dict[str, Dict[str, float]]:
    # Load CSV
    df = pd.read_csv(csv_path)

    # Define metric columns to analyze
    metric_columns = [
        "sequence", "jaccard", "levenshtein", "structure",
        "content", "content2", "f1", "similarity"
    ]

    # 1. Overall averages
    overall_averages = df[metric_columns].mean().round(4).to_dict()

    # 2. Language-level averages
    language_averages = (
        df.groupby("Language")[metric_columns]
        .mean()
        .round(4)
        .to_dict(orient="index")
    )

    return {
        "overall": overall_averages,
        "by_language": language_averages
    }
```

```python
import pandas as pd
from IPython.display import display

def display_stats(stats: dict, name: str = "Tool"):
    print(f"Evaluation Summary for: {name}\n")

    # Overall average metrics
    overall_df = pd.DataFrame.from_dict(stats["overall"], orient="index", columns=["Overall Average"])
    display(overall_df)

    # Language-level average metrics
    language_df = pd.DataFrame.from_dict(stats["by_language"], orient="index")
    language_df.index.name = "Language"
    display(language_df)
```

```python
import pandas as pd

def calculate_averages(csv_path: str):
    """
    Reads the evaluation CSV and calculates:
    - Overall average content2
    - Per-language average content2
    
    Args:
        csv_path (str): Path to the input CSV file.
    """
    # Load the CSV
    df = pd.read_csv(csv_path)

    # Make sure 'content2' is numeric
    df['content2'] = pd.to_numeric(df['content2'], errors='coerce')

    # Calculate overall average
    overall_avg = df['content2'].mean()

    # Calculate per-language averages
    language_avg = df.groupby('Language')['content2'].mean().sort_index()

    print(f"Overall Average content2: {overall_avg:.4f}\n")
    print("Per-Language Averages:")
    print(language_avg.round(4))

    return overall_avg, language_avg
```

```python
import pandas as pd
from pathlib import Path

def merge_content2_scores(
    spellcorrected_path: Path,
    pymupdf_path: Path,
    output_path: Path
):
    # Load the spell-corrected content2 scores
    df_spell = pd.read_csv(spellcorrected_path)
    df_spell.rename(columns={"content2": "content2_json_spellCorrected"}, inplace=True)

    # Load the PyMuPDF evaluation results
    df_pymupdf = pd.read_csv(pymupdf_path)
    df_pymupdf.rename(columns={"content2_json": "content2_json", "content2_html": "content2_pymupdf"}, inplace=True)

    # Merge the two datasets on CELEX ID and Language
    merged = pd.merge(df_spell, df_pymupdf, on=["CELEX ID", "Language"], how="inner")

    # Select and reorder desired columns
    merged = merged[[
        "Filepath", "CELEX ID", "Language",
        "content2_json_spellCorrected", "content2_json", "content2_pymupdf"
    ]]

    # Save the final merged result
    merged.to_csv(output_path, index=False)
    print(f"✅ Merged CSV saved to: {output_path}")
```

```python
# Usage example
merge_content2_scores(
    spellcorrected_path=Path("/ltstorage/home/4baba/EUR_lex/content2_scores.csv"),
    pymupdf_path=Path("/ltstorage/home/4baba/EUR_lex/evaluation_results/comparison_json_pymupdf.csv"),
    output_path=Path("/ltstorage/home/4baba/EUR_lex/merged_content2_comparison.csv")
)
```
